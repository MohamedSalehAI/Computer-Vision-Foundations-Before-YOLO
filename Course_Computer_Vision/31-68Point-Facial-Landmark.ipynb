# Real-Time Face Feature Detection Using 68 Facial Landmarks

```python
import dlib
import cv2
import time
import os
os.chdir(r"C:\Users\DT1\Desktop\proj-python\models")
```

```python

detector=dlib.get_frontal_face_detector()
predector=dlib.shape_predictor("shape_predictor_68_face_landmarks.dat")

prev_time=time.time()
right_eye=[]
left_eye=[]
eyebrows_right=[]
eyebrows_left=[]
nose=[]
mouth=[]
lip=[]
cheek_right=[]
cheek_left=[]

cap=cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH,640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT,480)
cap.set(cv2.CAP_PROP_FPS,30)
def puttext(frame,text,array,offset=(0,0)):
             x=int(sum([p[0] for p in array])/len(array))
             y=int(sum([p[1] for p in array])/len(array))
             (w,h),_=cv2.getTextSize(text,cv2.FONT_HERSHEY_COMPLEX,0.5,1)
             text_x=x-w//2
             text_y=y+h//2
            
             cv2.circle(frame,(x,y),2,(0,255,0),-1)
             font=cv2.FONT_HERSHEY_COMPLEX
             color=(255,0,0)
             thickness=2
             scale=0.4
             if offset==(0,0):
                 cv2.putText(frame,text,(text_x,text_y),font,scale,color,thickness)
             else:
                 cv2.putText(frame,text,(text_x+offset[0],text_y+offset[1]),font,scale,color,thickness)
while True:
    ret,frame=cap.read()
    if not ret:
        break
    current_time=time.time()
    fps=1/(current_time-prev_time)
    prev_time=current_time
    cv2.putText(frame,f"FPS:{int(fps)}",(20,50),cv2.FONT_HERSHEY_COMPLEX,1,(0,255,0),2)
    gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    blur=cv2.GaussianBlur(gray,(5,5),0)
    canny=cv2.Canny(blur,100,200)
    faces=detector(gray)
    for face in faces:
        x1=face.left()
        y1=face.top()
        x2=face.right()
        y2=face.bottom()
        landmark=predector(gray,face)
        for n in range(1,9):
            x=landmark.part(n).x
            y=landmark.part(n).y
            cheek_left.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
         
        for n in range(9,18):
            x=landmark.part(n).x
            y=landmark.part(n).y
            cheek_right.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
        
        for n in range(36,42):
            x=landmark.part(n).x
            y=landmark.part(n).y
            left_eye.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
        
        for n in range(42,48):
            x=landmark.part(n).x
            y=landmark.part(n).y
            right_eye.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
        
        for n in range(17,23):
            x=landmark.part(n).x
            y=landmark.part(n).y
            eyebrows_left.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
       
        for n in range(23,27):
            x=landmark.part(n).x
            y=landmark.part(n).y
            eyebrows_right.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
         
        for n in range(27,36):
            x=landmark.part(n).x
            y=landmark.part(n).y
            nose.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
         
        for n in range(48,60):
            x=landmark.part(n).x
            y=landmark.part(n).y
            mouth.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
         
        for n in range(60,68):
            x=landmark.part(n).x
            y=landmark.part(n).y
            lip.append((x,y))
            cv2.circle(frame,(x,y),2,(0,255,0),-1)
         
        cv2.rectangle(frame,(x1,y1),(x2,y2),(0,255,0),2)
        center_face=(int((x1+x2)/2),int((y1+y2)/2))
        cv2.putText(frame,"face",(center_face[0],center_face[1]),cv2.FONT_HERSHEY_COMPLEX,0.5,(0,0,255),1)
        
        puttext(frame,"left_eye",left_eye,(-10,0))
        puttext(frame,"right_eye",right_eye,(20,0))
        puttext(frame,"eyebrows_left",eyebrows_left,(-10,0))
        puttext(frame,"eyebrows_right",eyebrows_right,(20,0))
        puttext(frame,"nose",nose)
        puttext(frame,"mouth",mouth,(0,-10))
        puttext(frame,"lip",lip,(0,10))
        puttext(frame,"cheek_left",cheek_left,(10,0))
        puttext(frame,"cheek_right",cheek_right,(20,0))
          
        
 
    cv2.imshow("frame",frame)
    cv2.imshow("canny",canny)
    if cv2.waitKey(1)==ord("q"):
        break
cap.release()
cv2.destroyAllWindows()
            
```
