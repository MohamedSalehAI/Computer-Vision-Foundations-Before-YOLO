```python
import keras_ocr
import matplotlib.pyplot as plt

# ==========================================
# 1. Create Keras-OCR pipeline
# ==========================================

pipeline = keras_ocr.pipeline.Pipeline()

print("Keras-OCR pipeline loaded successfully!")


# ==========================================
# 2. Read image
# ==========================================

 

images = [keras_ocr.tools.read(img) for img in ["street_sign1.jpg","street_sign2.jpg","street_sign3.jpg","street_sign4.jpg","street_sign5.jpg","street_sign6.jpg"]]


# ==========================================
# 3. Recognize text
# ==========================================

prediction_groups = pipeline.recognize(images)
print("prediction_groups:", prediction_groups)
predictions = prediction_groups[0]


# ==========================================
# 4. Print detected text
# ==========================================

print("\nDetected text:")
print("-" * 40)

for text, box in predictions:
    print(text)


# ==========================================
# 5. Display detected text + bounding boxes
# ==========================================

fig, ax = plt.subplots(nrows=len(images), figsize=(15, 15))

for i, (image, predictions) in enumerate(zip(images, prediction_groups)):
    keras_ocr.tools.drawAnnotations(
        image=image,
        predictions=predictions,
        ax=ax[i]
    )

plt.title("Keras-OCR Text Detection and Recognition")
plt.axis("off")
plt.tight_layout()
plt.show()
```
