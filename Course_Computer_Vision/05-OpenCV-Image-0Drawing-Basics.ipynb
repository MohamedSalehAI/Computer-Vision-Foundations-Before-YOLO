```python
import cv2
import numpy as np

# Create image
img = np.zeros((840, 840, 3), dtype=np.uint8)#(height, width, channels)numpy

# Sky
cv2.rectangle(img, (0, 0), (840, 600), (255, 220, 150),-1)#(width,hieght)opencv

# Ground
cv2.rectangle(img, (0, 600), (840, 840), (40, 150, 40), -1)

# Sun
cv2.circle(img, (650, 150), 80, (0, 220, 255), -1)

# Mountains
mountain = np.array([
    [0, 600],
    [250, 350],
    [450, 600]
])

cv2.fillPoly(img, [mountain], (100, 100, 100))

mountain2 = np.array([
    [350, 600],
    [600, 300],
    [840, 600]
])

cv2.fillPoly(img, [mountain2], (80, 80, 80))

# Main tree trunk
cv2.line(img, (420, 600), (420, 400), (40, 30, 20), 12)

# Main tree leaves
cv2.circle(img, (420, 370), 80, (0, 130, 0), -1)
cv2.circle(img, (360, 420), 60, (0, 150, 0), -1)
cv2.circle(img, (480, 420), 60, (0, 150, 0), -1)

# Small tree
cv2.line(img, (180, 600), (180, 450), (40, 30, 20), 8)

cv2.circle(img, (180, 420), 60, (0, 130, 0), -1)
cv2.circle(img, (140, 450), 45, (0, 150, 0), -1)
cv2.circle(img, (220, 450), 45, (0, 150, 0), -1)

# Clouds
cv2.circle(img, (150, 150), 30, (255, 255, 255), -1)
cv2.circle(img, (190, 140), 40, (255, 255, 255), -1)
cv2.circle(img, (230, 150), 30, (255, 255, 255), -1)

# Text
font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(
    img,
    "EARTH",
    (350, 70),
    font,
    1.5,
    (0, 0, 0),
    3,
    cv2.LINE_AA
)

# Show image
cv2.imshow("Beautiful Earth", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# saving image

```python
cv2.imwrite("earth.png", img)
cv2.imshow("Earth", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Simple Earth drawing project

```python
import cv2
import numpy as np

# ==============================
# Create image
# ==============================
img = np.zeros((840, 840, 3), dtype=np.uint8)

# ==============================
# Sky
# ==============================
img[:600] = (220, 180, 100)   # sky

# Add several horizontal layers to make sky prettier
for y in range(0, 600, 20):
    color = (
        max(100, 220 - y // 8),
        max(120, 180 - y // 10),
        max(180, 100 + y // 5)
    )
    cv2.rectangle(img, (0, y), (840, y + 20), color, -1)

# ==============================
# Ground
# ==============================
cv2.rectangle(img, (0, 600), (840, 840), (40, 120, 40), -1)

# Ground layers
cv2.rectangle(img, (0, 600), (840, 620), (30, 100, 30), -1)

# ==============================
# Sun
# ==============================

# Sun glow
cv2.circle(img, (650, 180), 130, (180, 220, 255), -1)

# Sun
cv2.circle(img, (650, 180), 90, (0, 220, 255), -1)

# Inner sun
cv2.circle(img, (650, 180), 70, (0, 240, 255), -1)

# ==============================
# Clouds
# ==============================

def draw_cloud(image, x, y):
    white = (245, 245, 245)

    cv2.circle(image, (x, y), 35, white, -1)
    cv2.circle(image, (x + 40, y - 15), 45, white, -1)
    cv2.circle(image, (x + 85, y), 35, white, -1)
    cv2.rectangle(
        image,
        (x - 10, y),
        (x + 95, y + 30),
        white,
        -1
    )

draw_cloud(img, 120, 160)
draw_cloud(img, 330, 100)

# ==============================
# Mountains
# ==============================

mountain1 = np.array([
    [0, 600],
    [130, 420],
    [250, 600]
], np.int32)

mountain2 = np.array([
    [180, 600],
    [400, 380],
    [620, 600]
], np.int32)

mountain3 = np.array([
    [500, 600],
    [680, 430],
    [840, 600]
], np.int32)

cv2.fillPoly(img, [mountain1], (80, 90, 100))
cv2.fillPoly(img, [mountain2], (100, 110, 120))
cv2.fillPoly(img, [mountain3], (70, 80, 90))

# Snow on mountains
snow1 = np.array([
    [400, 380],
    [360, 430],
    [390, 415],
    [400, 440],
    [430, 410],
    [450, 430]
], np.int32)

cv2.fillPoly(img, [snow1], (240, 240, 240))

# ==============================
# Main Tree
# ==============================

# Tree trunk
cv2.line(
    img,
    (420, 600),
    (420, 390),
    (50, 40, 20),
    18
)

# Branches
cv2.line(img, (420, 470), (360, 420), (50, 40, 20), 10)
cv2.line(img, (420, 450), (480, 400), (50, 40, 20), 10)

# Tree leaves
cv2.circle(img, (360, 410), 65, (20, 120, 20), -1)
cv2.circle(img, (420, 370), 75, (20, 150, 30), -1)
cv2.circle(img, (480, 410), 65, (20, 130, 20), -1)
cv2.circle(img, (420, 430), 70, (30, 140, 25), -1)

# ==============================
# Small Tree
# ==============================

cv2.line(
    img,
    (180, 600),
    (180, 420),
    (50, 40, 20),
    12
)

cv2.circle(img, (140, 420), 45, (20, 130, 20), -1)
cv2.circle(img, (180, 390), 55, (20, 150, 30), -1)
cv2.circle(img, (220, 420), 45, (20, 130, 20), -1)

# ==============================
# Grass
# ==============================

for x in range(0, 840, 30):

    cv2.line(
        img,
        (x, 600),
        (x + 8, 585),
        (20, 80, 20),
        2
    )

# ==============================
# Title
# ==============================

font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(
    img,
    "EARTH",
    (300, 45),
    font,
    1.8,
    (25, 55, 255),
    2,
    cv2.LINE_AA
)

# Small subtitle
cv2.putText(
    img,
    "A beautiful world",
    (150, 260),
    font,
    1.6,
    (240, 40, 240),
    1,
    cv2.LINE_AA
)

# ==============================
# Display
# ==============================

cv2.imshow("Beautiful Earth", img)

key = cv2.waitKey(0)

if key == ord('q'):
    cv2.destroyAllWindows()
```
