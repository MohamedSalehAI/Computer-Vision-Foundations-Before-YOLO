# By using Arabic_Reshaper for Arabic_text

```python
import cv2
import numpy as np
from PIL import Image, ImageDraw, ImageFont
from bidi.algorithm import get_display
import arabic_reshaper
import os
os.chdir(r"D:\proj-python\Images_Vision")

# Load the image
img = cv2.imread("image_1.jpg")

# Load a font that supports Arabic characters
fontpath = "arial.ttf"  # Make sure to download an Arabic font and specify the correct path
font = ImageFont.truetype(fontpath, 32)

# Convert the image to a PIL image
img_pil = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

# Create a drawing context
draw = ImageDraw.Draw(img_pil)

# Arabic text
text = "اللغة العربية"

# Reshape and bidi the text
reshaped_text = arabic_reshaper.reshape(text)
bidi_text = get_display(reshaped_text)

# Position and font color
x, y = 100, 80
font_color = (255, 255, 255)  # White

# Draw the text on the image
draw.text((x, y), bidi_text, font=font, fill=font_color)

# Convert back to a NumPy array
img = np.array(img_pil)

# Display the image with Arabic text
cv2.imshow("Image with Arabic", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2
import numpy as np
from PIL import Image, ImageDraw, ImageFont

img = cv2.imread("image_1.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


font= ImageFont.truetype("arial.ttf", 50)
img_pil = Image.fromarray(img)
draw = ImageDraw.Draw(img_pil)

text = "اللغة العربية"

draw.text((100, 100), text, font=font, fill=(255, 0, 0), direction="rtl")

img = cv2.cvtColor(np.array(img_pil), cv2.COLOR_RGB2BGR)

cv2.imshow("Arabic", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

import cv2
import numpy as np

from PIL import Image, ImageDraw, ImageFont

import os

os.chdir(r"D:\proj-python\Images_Vision")

y_true = ["تفاحة", "تفاحة", "كلب", "كلب", "قطة", "قطة", "تفاحة", "كلب", "قطة", "تفاحة"]
y_pred = ["تفاحة", "كلب", "كلب", "كلب", "قطة", "تفاحة", "تفاحة", "قطة", "قطة", "تفاحة"]

class_names = ["تفاحة", "كلب", "قطة"]

cm = confusion_matrix(y_true, y_pred, labels=class_names)

plt.figure(figsize=(8, 6))

sns.heatmap(cm, annot=True, fmt="d", xticklabels=class_names, yticklabels=class_names)

plt.xlabel("التوقع")
plt.ylabel("الحقيقة")
plt.title("مصفوفة الالتباس")

plt.show()

```
