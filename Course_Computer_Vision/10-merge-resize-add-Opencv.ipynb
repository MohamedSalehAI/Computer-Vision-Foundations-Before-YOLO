```python
import cv2

img1 = cv2.imread(r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_1.jpg")
img2 = cv2.imread(r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_2.jpg")
img3 = cv2.imread(r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_3.jpg")
img4 = cv2.imread(r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_4.jpg")

# Resize all images
size = (400, 300)

img1 = cv2.resize(img1, size)
img2 = cv2.resize(img2, size)
img3 = cv2.resize(img3, size)
img4 = cv2.resize(img4, size)

# Create rows
row1 = cv2.hconcat([img1, img2])
row2 = cv2.hconcat([img3, img4])

# Create final collage
collage = cv2.vconcat([row1, row2])

cv2.imshow("Image Collage", collage)

cv2.imwrite("collage.jpg", collage)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
