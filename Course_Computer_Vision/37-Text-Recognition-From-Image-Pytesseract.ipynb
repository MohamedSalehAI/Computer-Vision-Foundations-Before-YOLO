|  PSM | Meaning                     |
| ---: | --------------------------- |
|  `3` | Automatic page segmentation |
|  `4` | Single column               |
|  `6` | One uniform block of text   |
|  `7` | Single text line            |
|  `8` | Single word                 |
|  `9` | Single word in a circle     |
| `10` | Single character            |
| `11` | Sparse text                 |
| `12` | Sparse text + OSD           |
| `13` | Raw single line             |


```python
model_path=r"C:\Program Files\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd=model_path
img=cv2.imread("text_recognition1.jpg")
img=cv2.resize(img,None,fx=1.5,fy=1.5)
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
gray=cv2.GaussianBlur(gray,(5,5),0)
thresh=cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,45,7)
# kernal=cv2.getStructuringElement(cv2.MORPH_RECT,(3,3))
# open=cv2.morphologyEx(thresh,cv2.MORPH_OPEN,kernal,iterations=1)
# close=cv2.morphologyEx(open,cv2.MORPH_CLOSE,kernal,iterations=1)
 
cv2.imshow("threshold",thresh)
conf="--psm 3"
text=pytesseract.image_to_string(thresh,lang="eng",config=conf)
print(text)

contours,hierarchy=cv2.findContours(thresh,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)

for cnt in contours:
    x,y,w,h=cv2.boundingRect(cnt)
    area=w*h
    if area>300 and w>20 and h>20:
        roi = thresh[y:y+h, x:x+w]  
        text = pytesseract.image_to_string(roi, lang="eng", config=conf).strip()
        # print(f"Text: {text.strip()}")qq
        if text:
           conf="--psm 7"
           cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),1)
           cv2.putText(img,text,(x,y-10),cv2.FONT_HERSHEY_SIMPLEX,0.5,(0,0,255),1)

cv2.imshow("img",img)
key =cv2.waitKey(0)& 0xFF
if key ==ord('q'):
   cv2.destroyAllWindows()
```
