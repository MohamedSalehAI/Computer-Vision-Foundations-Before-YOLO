# apply concepts on simple project

```python
import cv2
import datetime
import time

# ---------------------------------------
# Camera
# ---------------------------------------
cap = cv2.VideoCapture(0)

cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)
cap.set(cv2.CAP_PROP_FPS, 80)

if not cap.isOpened():
    print("Camera could not be opened.")
    exit()

# ---------------------------------------
# FPS calculation
# ---------------------------------------
prev_time = time.time()
frame_count = 0

# ---------------------------------------
# Font
# ---------------------------------------
font = cv2.FONT_HERSHEY_SIMPLEX

while True:

    ret, frame = cap.read()

    if not ret:
        print("Failed to read frame.")
        break

    frame_count += 1

    # -----------------------------------
    # Calculate real FPS
    # -----------------------------------
    current_time = time.time()
    fps = 1 / (current_time - prev_time)
    prev_time = current_time

    # -----------------------------------
    # Information
    # -----------------------------------
    height, width = frame.shape[:2]

    timestamp = datetime.datetime.now().strftime("%H:%M:%S")

    # -----------------------------------
    # Dark information panel
    # -----------------------------------
    overlay = frame.copy()

    cv2.rectangle(
        overlay,
        (15, 15),
        (360, 125),
        (20, 20, 20),
        -1
    )

    # Transparency
    frame = cv2.addWeighted(
        overlay,
        0.65,
        frame,
        0.35,
        0
    )

    # -----------------------------------
    # Text
    # -----------------------------------

    cv2.putText(
        frame,
        "CAMERA MONITOR",
        (30, 45),
        font,
        0.7,
        (255, 255, 255),
        2,
        cv2.LINE_AA
    )

    cv2.putText(
        frame,
        f"{width} x {height}",
        (30, 75),
        font,
        0.55,
        (220, 220, 220),
        1,
        cv2.LINE_AA
    )

    cv2.putText(
        frame,
        f"FPS  {fps:.1f}",
        (30, 102),
        font,
        0.55,
        (220, 220, 220),
        1,
        cv2.LINE_AA
    )

    # -----------------------------------
    # Timestamp
    # -----------------------------------
    cv2.putText(
        frame,
        timestamp,
        (width - 150, 35),
        font,
        0.6,
        (255, 255, 255),
        1,
        cv2.LINE_AA
    )

    # -----------------------------------
    # Bottom status
    # -----------------------------------
    cv2.putText(
        frame,
        "Press Q to exit",
        (30, height - 25),
        font,
        0.5,
        (20, 30, 230),
        1,
        cv2.LINE_AA
    )

    # -----------------------------------
    # Display
    # -----------------------------------
    cv2.imshow("Camera Monitor", frame)

    # -----------------------------------
    # Exit
    # -----------------------------------
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# ---------------------------------------
# Cleanup
# ---------------------------------------
cap.release()
cv2.destroyAllWindows()

print(f"Frames captured: {frame_count}")
```
