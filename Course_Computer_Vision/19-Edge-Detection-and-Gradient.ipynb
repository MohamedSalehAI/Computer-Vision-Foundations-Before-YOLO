# Simple Project

```python
import cv2
import numpy as np


# ============================================================
# Image Gradient & Edge Detection using Laplacian
# ============================================================


# ------------------------------------------------------------
# 1. Load Image
# ------------------------------------------------------------

img = cv2.imread("sudoku.jpg")

if img is None:
    print("ERROR: Could not load image.")
    print("Make sure test_image.jpg is in the same folder.")
    exit()


# Resize image for display
img = cv2.resize(img, (400, 300))


# ------------------------------------------------------------
# 2. Convert to Grayscale
# ------------------------------------------------------------

gray = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2GRAY
)


# ------------------------------------------------------------
# 3. Add Title
# ------------------------------------------------------------

def add_title(image, title):

    # Convert grayscale image to BGR
    if len(image.shape) == 2:
        image = cv2.cvtColor(
            image,
            cv2.COLOR_GRAY2BGR
        )

    height, width = image.shape[:2]

    # Header
    header = np.ones(
        (45, width, 3),
        dtype=np.uint8
    ) * 255

    cv2.putText(
        header,
        title,
        (10, 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.65,
        (0, 0, 0),
        2,
        cv2.LINE_AA
    )

    return cv2.vconcat([
        header,
        image
    ])


# ------------------------------------------------------------
# 4. Trackbar Callback
# ------------------------------------------------------------

def nothing(x):
    pass


# ------------------------------------------------------------
# 5. Create Window
# ------------------------------------------------------------

cv2.namedWindow(
    "Laplacian Edge Analyzer"
)


# Gaussian Kernel
cv2.createTrackbar(
    "Blur Kernel",
    "Laplacian Edge Analyzer",
    1,
    10,
    nothing
)


# Laplacian Kernel
cv2.createTrackbar(
    "Laplacian Kernel",
    "Laplacian Edge Analyzer",
    1,
    2,
    nothing
)


# Threshold
cv2.createTrackbar(
    "Threshold",
    "Laplacian Edge Analyzer",
    30,
    255,
    nothing
)


# Canny lower threshold
cv2.createTrackbar(
    "Canny Low",
    "Laplacian Edge Analyzer",
    50,
    255,
    nothing
)


# Canny upper threshold
cv2.createTrackbar(
    "Canny High",
    "Laplacian Edge Analyzer",
    150,
    255,
    nothing
)


# ------------------------------------------------------------
# 6. Main Loop
# ------------------------------------------------------------

while True:

    # ========================================================
    # Read Trackbar Values
    # ========================================================

    blur_value = cv2.getTrackbarPos(
        "Blur Kernel",
        "Laplacian Edge Analyzer"
    )

    laplace_value = cv2.getTrackbarPos(
        "Laplacian Kernel",
        "Laplacian Edge Analyzer"
    )

    threshold = cv2.getTrackbarPos(
        "Threshold",
        "Laplacian Edge Analyzer"
    )

    canny_low = cv2.getTrackbarPos(
        "Canny Low",
        "Laplacian Edge Analyzer"
    )

    canny_high = cv2.getTrackbarPos(
        "Canny High",
        "Laplacian Edge Analyzer"
    )


    # ========================================================
    # Make Gaussian Kernel Odd
    # ========================================================

    blur_kernel = blur_value * 2 + 1


    # ========================================================
    # Laplacian Kernel
    # ========================================================

    # 1 -> 3x3
    # 2 -> 5x5

    laplace_kernel = laplace_value * 2 + 1


    # ========================================================
    # 7. Gaussian Blur
    # ========================================================

    blurred = cv2.GaussianBlur(
        gray,
        (blur_kernel, blur_kernel),
        0
    )


    # ========================================================
    # 8. Laplacian
    # ========================================================

    laplacian = cv2.Laplacian(
        blurred,
        cv2.CV_64F,
        ksize=laplace_kernel
    )


    # ========================================================
    # 9. Convert Laplacian to Displayable Image
    # ========================================================

    laplacian_abs = cv2.convertScaleAbs(
        laplacian
    )


    # ========================================================
    # 10. Sobel X
    # ========================================================

    sobel_x = cv2.Sobel(
        blurred,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    sobel_x = cv2.convertScaleAbs(
        sobel_x
    )


    # ========================================================
    # 11. Sobel Y
    # ========================================================

    sobel_y = cv2.Sobel(
        blurred,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    sobel_y = cv2.convertScaleAbs(
        sobel_y
    )


    # ========================================================
    # 12. Sobel Gradient Magnitude
    # ========================================================

    sobel_magnitude = cv2.magnitude(
        sobel_x.astype(np.float32),
        sobel_y.astype(np.float32)
    )

    sobel_magnitude = cv2.convertScaleAbs(
        sobel_magnitude
    )


    # ========================================================
    # 13. Laplacian Threshold
    # ========================================================

    _, laplacian_binary = cv2.threshold(
        laplacian_abs,
        threshold,
        255,
        cv2.THRESH_BINARY
    )


    # ========================================================
    # 14. Canny Edge Detection
    # ========================================================

    if canny_high <= canny_low:
        canny_high = canny_low + 1

    canny = cv2.Canny(
        blurred,
        canny_low,
        canny_high
    )


    # ========================================================
    # 15. Add Titles
    # ========================================================

    original_view = add_title(
        img,
        "Original"
    )

    gray_view = add_title(
        gray,
        "Grayscale"
    )

    blurred_view = add_title(
        blurred,
        f"Gaussian {blur_kernel}x{blur_kernel}"
    )

    laplacian_view = add_title(
        laplacian_abs,
        f"Laplacian {laplace_kernel}x{laplace_kernel}"
    )

    sobel_x_view = add_title(
        sobel_x,
        "Sobel X"
    )

    sobel_y_view = add_title(
        sobel_y,
        "Sobel Y"
    )

    sobel_mag_view = add_title(
        sobel_magnitude,
        "Sobel Magnitude"
    )

    laplacian_binary_view = add_title(
        laplacian_binary,
        f"Laplacian Threshold {threshold}"
    )

    canny_view = add_title(
        canny,
        f"Canny {canny_low}-{canny_high}"
    )


    # ========================================================
    # 16. Create Rows
    # ========================================================

    row1 = cv2.hconcat([
        original_view,
        gray_view,
        blurred_view
    ])

    row2 = cv2.hconcat([
        laplacian_view,
        sobel_x_view,
        sobel_y_view
    ])

    row3 = cv2.hconcat([
        sobel_mag_view,
        laplacian_binary_view,
        canny_view
    ])


    # ========================================================
    # 17. Create Final Dashboard
    # ========================================================

    dashboard = cv2.vconcat([
        row1,
        row2,
        row3
    ])


    # ========================================================
    # 18. Display
    # ========================================================

    cv2.imshow(
        "Laplacian Edge Analyzer",
        dashboard
    )


    # ========================================================
    # 19. Keyboard Controls
    # ========================================================

    key = cv2.waitKey(30) & 0xFF


    # Quit
    if key == ord("q"):
        break


    # Save
    elif key == ord("s"):

        cv2.imwrite(
            "laplacian_edge_result.jpg",
            dashboard
        )

        print(
            "Result saved as "
            "laplacian_edge_result.jpg"
        )


# ------------------------------------------------------------
# 20. Cleanup
# ------------------------------------------------------------

cv2.destroyAllWindows()
```
