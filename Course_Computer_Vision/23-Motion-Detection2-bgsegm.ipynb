```python
min_h=80
min_w=80
offset=5

detect=[]
cars=0
def calc_center(x,y,w,h):
    cx=x+w//2
    cy=y+h//2
    return cx,cy
cap=cv2.VideoCapture("cars_motion.mp4")
tracker = cv2.bgsegm.createBackgroundSubtractorMOG(history=50)
h=int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
w=int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
print("height",h)
print("width",w)
pos_line=int(h * 0.79)
print("pos_line",pos_line)
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    gray=cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blur=cv2.GaussianBlur(gray, (5,5), 2)
    mask=tracker.apply(blur)
    dilat=cv2.dilate(mask, np.ones((7,7)))
    # kernal=cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
    # opening=cv2.morphologyEx(dilat, cv2.MORPH_OPEN, kernal, iterations=2)
    # closing=cv2.morphologyEx(dilat, cv2.MORPH_CLOSE, kernal, iterations=2)
    contours,hiracky=cv2.findContours(dilat,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)
    cv2.drawContours(frame, contours, -1, (0,255,0), 2)
    cv2.line(frame,(0,pos_line),(640,pos_line),(255,0,0),3)
    for id,cont in enumerate(contours):
        (x,y,w,h)=cv2.boundingRect(cont)
        if w>min_w and h>min_h:
            cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),2)
            cx,cy=calc_center(x,y,w,h)
            cv2.circle(frame,(cx,cy),5,(0,0,255),-1)
            detect.append((cx,cy))
        for (x,y) in detect:
            if y<(pos_line+offset) and y>(pos_line-offset):
                cars+=1
                cv2.line(frame,(0,pos_line),(640,pos_line),(0,0,255),3)
                detect.remove((x,y))
                print(f"Car count: {cars}")



    cv2.putText(frame,f"Counter of Cars: {cars}",(10,50),cv2.FONT_HERSHEY_SIMPLEX,1,(0,0,255),2)
    cv2.imshow("Original Frame", frame)
    # cv2.imshow("Mask", mask)
    # cv2.imshow("Detector Cars",dilat)
    # cv2.imshow("closing",closing)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
     
    
```
