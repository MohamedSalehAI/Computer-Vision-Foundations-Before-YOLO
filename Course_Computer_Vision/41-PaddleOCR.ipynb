```python
import cv2
from paddleocr import PaddleOCR

img = cv2.imread("street_sign1.jpg")

ocr = PaddleOCR(lang="en", device="cpu", enable_mkldnn=False)
result = ocr.predict(img)

for res in result:
    data = res.json["res"]

    for box, text, score in zip(
        data["rec_boxes"],
        data["rec_texts"],
        data["rec_scores"]
    ):
        if score < 0.5:
            continue

        x1, y1, x2, y2 = map(int, box)

        print(text, "confidence:", round(float(score), 2))

        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(img, text, (x1, y1 - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

cv2.imshow("OCR", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.figure(figsize=(10,10))
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title("image text english paddlecor")
plt.show()
```

# For HandWriting in Arabic

```python
img=cv2.imread("handwriting_arabic.jpg")
ocr=PaddleOCR(lang="ar", device="cpu", enable_mkldnn=False)
result = ocr.predict(img)

for res in result:
    print(res)
    data=res.json["res"]
    print(res.json.keys())
    texts=data["rec_texts"]
    blocks=data["rec_boxes"]
    scores=data["rec_scores"]
    for box, text, score in zip(blocks,texts,scores):
        if score < 0.5:
            continue

        x1, y1, x2, y2 = map(int, box)

        print(text, "confidence:", round(float(score), 2))

        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(img, text, (x1, y1 - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

cv2.imshow("OCR", img)
cv2.waitKey(0)
cv2.destroyAllWindows()

plt.figure(figsize=(10,10))
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title("image text english paddlecor")
plt.show()
```
