# function of CLick_Event

```python
def click_event(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        print(x, ',', y)
        font = cv2.FONT_HERSHEY_SIMPLEX
        strXY = str(x) + ', ' + str(y)
        cv2.putText(img, strXY, (x, y), font, 1, (255, 255, 0), 2)
        cv2.imshow('image', img)
    elif event==cv2.EVENT_RBUTTONDOWN:
        print(x, ',', y)
        font = cv2.FONT_HERSHEY_SIMPLEX
        strXY = str(x) + ', ' + str(y)
        cv2.putText(img, strXY, (x, y), font, 1, (255, 255, 0), 2)
        cv2.imshow('image', img)
        blue=img[y,x,0]
        green=img[y,x,1]
        red=img[y,x,2]
        cv2.circle(img,(x,y),3,(0,0,255),-1)
        mycolorimage=np.zeros((512,512,3),np.uint8)
        mycolorimage[:]=[blue,green,red]
        cv2.imshow('color',mycolorimage)
```

# Simple project eventa

```python
from sched import Event
import cv2
import numpy as np

img=255*np.ones((512,512,3),dtype=np.uint8)

def mouse_event(event,x,y,flags,param):
    if event==cv2.EVENT_LBUTTONDOWN:
        print(f"Left Click: {x},{y}")
        cv2.circle(img,(x,y),15,(0,158,15),-1)
    elif event==cv2.EVENT_RBUTTONDOWN:
        print(f"Right Click: {x},{y}")
        cv2.circle(img, (x, y), 8, (255, 0, 0), -1)
    elif event==cv2.EVENT_MOUSEMOVE:
        if flags&cv2.EVENT_FLAG_LBUTTON:
            cv2.circle(img, (x, y), 3, (0, 0, 0), -1)
# Create window
cv2.namedWindow("Mouse Event")

# Connect mouse callback
cv2.setMouseCallback("Mouse Event", mouse_event)

while True:

    cv2.imshow("Mouse Event", img)

    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break

cv2.destroyAllWindows()
```
