```python
import cv2
import os
os.chdir(r"C:\Users\DT1\Desktop\proj-python\Images_Vision")

# Read image
img = cv2.imread("logo_eagle.jpg")

# Resize
img = cv2.resize(img, (800, 600))

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Threshold
_, binary = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

# Find contours
contours, hierarchy = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Number of objects:", len(contours))

# Draw contours
result = img.copy()

cv2.drawContours(
    result,
    contours,
    -1,
    (0, 255, 0),
    2
)

# Put object number on image
for i, contour in enumerate(contours):

    # Calculate center
    M = cv2.moments(contour)

    if M["m00"] != 0:
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])

        cv2.putText(
            result,
            f"Object {i + 1}",
            (cx - 40, cy),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 0, 255),
            2
        )

# Display
cv2.imshow("Original", img)
cv2.imshow("Binary", binary)
cv2.imshow("Contours", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
