# Otsu calculates the optimal threshold from the image histogram

```python
img=cv2.imread("noise_image.jpg",0)
ret,th1=cv2.threshold(img,127,255,cv2.THRESH_BINARY)
ret,th2=cv2.threshold(img,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
blur=cv2.GaussianBlur(img,(5,5),0)
ret,th3=cv2.threshold(blur,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
images=[img,0,th1,
img,0,th2,
blur,0,th3]
titles=["Original image","histogram","Global Threshold",
"Original image","histogram","THRESH_OTSU Threshold",
"Blur image","histogram","THRESH_OTSU+BLUR"]
plt.figure(figsize=(15,12))
for i in range(3):
    plt.subplot(3,3,i*3+1)
    plt.imshow(images[i*3],"gray")
    plt.xticks([]),plt.yticks([])
    plt.title(titles[i*3])
 

    plt.subplot(3,3,i*3+2)
    plt.hist(images[i*3].ravel(),256,[0,256])
    plt.xticks([]),plt.yticks([])
    plt.title(titles[i*3+1])

    plt.subplot(3,3,i*3+3)
    plt.imshow(images[i*3+2],"gray")
    plt.xticks([]),plt.yticks([])
    plt.title(titles[i*3+2])
plt.show()
```
