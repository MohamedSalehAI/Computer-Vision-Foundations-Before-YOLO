```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os
os.chdir(r"C:\Users\DT1\Downloads")
```

```python
img=cv2.imread("box_balls.jpg",0)
_,mask=cv2.threshold(img,220,255,cv2.THRESH_BINARY_INV)

kernal=np.ones((5,5),dtype=np.uint8)
dilation=cv2.dilate(mask,kernal,iterations=2)
erosion=cv2.erode(mask,kernal,iterations=3)
opening=cv2.morphologyEx(mask,cv2.MORPH_OPEN,kernal,iterations=4)
closing=cv2.morphologyEx(mask,cv2.MORPH_CLOSE,kernal,iterations=4)

cv2.imshow("original image",img)
cv2.imshow("mask image",mask)
cv2.imshow("dilation image",dilation)
cv2.imshow("erosion image",erosion)
cv2.imshow("opeing image",opening)
cv2.imshow("closing image",closing)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
