```python
from PIL import Image

img = Image.open("image_1.jpg")

resized = img.resize((640, 480))

print("Original:", img.size)
print("Resized:", resized.size)

pixel = img.getpixel((100, 50))

print("Pixel at (100, 50):", pixel)
R,G,B = pixel
print("R:", R, "G:", G, "B:", B)
```

```python
import cv2
from PIL import Image
import matplotlib.pyplot as plt
image = cv2.imread("image_1.jpg")
h,w,c=image.shape
print("height:", h)
print("width:", w)
print("channels:", c)
print("dtype:", image.dtype)
print("shape:", image.shape)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

pil_img = Image.fromarray(image_rgb)
pil_img.show()
w,h=pil_img.size
print("width:", w)
print("height:", h)
print("mode:", pil_img.mode)
plt.figure(figsize=(15,15))
plt.imshow(pil_img)
plt.axis("off")
plt.show()
```

```python
from PIL import Image,ImageDraw
import matplotlib.pyplot as plt

img = Image.open("image_2.jpg")

draw = ImageDraw.Draw(img)

draw.rectangle(
    (100, 100, 500, 400),
    outline="red",
    width=5
)

img.show()
# img.save("output.jpg")
plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axis("off")
plt.show()
img.close()
```

```python
from PIL import Image, ImageEnhance, ImageFilter
import numpy as np
import matplotlib.pyplot as plt

# 1. Read
img = Image.open("image_3.jpg")

print("Original:")
print("Size:", img.size)
print("Mode:", img.mode)

# 2. Resize
img = img.resize((640, 480))

# 3. Crop ROI
roi = img.crop((100, 100, 500, 400))

# 4. Convert to grayscale
gray = roi.convert("L")

# 5. Blur
blur = gray.filter(ImageFilter.GaussianBlur(2))

# 6. Convert to NumPy
array = np.array(blur)

print("ROI:", roi.size)
print("Array shape:", array.shape)
print("Data type:", array.dtype)

# 7. Display
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(roi)
plt.title("Original ROI")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(array, cmap="gray")
plt.title("Grayscale + Blur")
plt.axis("off")

plt.show()
```
