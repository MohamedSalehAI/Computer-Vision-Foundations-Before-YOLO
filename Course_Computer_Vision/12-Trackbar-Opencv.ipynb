```python
import cv2
import numpy as np


# ==========================================
# Create Window
# ==========================================

window = "RGB Color Mixer"

cv2.namedWindow(window)


# ==========================================
# Trackbar Callback
# ==========================================

def nothing(value):
    pass


# ==========================================
# Create RGB Trackbars
# ==========================================

cv2.createTrackbar(
    "Blue",
    window,
    0,
    255,
    nothing
)

cv2.createTrackbar(
    "Green",
    window,
    0,
    255,
    nothing
)

cv2.createTrackbar(
    "Red",
    window,
    0,
    255,
    nothing
)


# ==========================================
# Main Loop
# ==========================================

while True:

    # Get trackbar values
    blue = cv2.getTrackbarPos("Blue", window)
    green = cv2.getTrackbarPos("Green", window)
    red = cv2.getTrackbarPos("Red", window)

    # Create image with selected color
    img = np.zeros(
        (512, 512, 3),
        dtype=np.uint8
    )

    # OpenCV uses BGR order
    img[:] = [blue, green, red]

    # Display color
    cv2.imshow(window, img)

    # Press Q to exit
    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break


# ==========================================
# Cleanup
# ==========================================

cv2.destroyAllWindows()
```
