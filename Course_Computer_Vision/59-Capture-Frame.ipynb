```python

cap=cv2.VideoCapture(1)

prev_time=time.time()
frame_counter=0
cap_frame=0
number_img_captured=20
capture_img=False

Dir_name="Capturing_img"

if not os.path.exists(Dir_name):
    os.makedirs(Dir_name)

while True:

    ret,frame=cap.read()

    if not ret:
        print("Failed to capture frame")
        break
    frame=cv2.rotate(frame,cv2.ROTATE_90_CLOCKWISE)

    frame_counter+=1

    h,w,c=frame.shape

    current_time=time.time()
    fps=frame_counter/(current_time-prev_time)

    cv2.putText(frame,f'FPS:{int(fps)}',(20,40),cv2.FONT_HERSHEY_PLAIN,2,(0,255,0),2)
    cv2.putText(frame,f'Height:{h}',(20,80),cv2.FONT_HERSHEY_PLAIN,2,(0,255,0),2)
    cv2.putText(frame,f'Width:{w}',(20,120),cv2.FONT_HERSHEY_PLAIN,2,(0,255,0),2)

    if capture_img and cap_frame<number_img_captured:

        cap_frame+=1

        cv2.imwrite(f'{Dir_name}/img_{cap_frame}.jpg',frame)

        cv2.putText(frame,"Capturing",(20,450),cv2.FONT_HERSHEY_PLAIN,2,(0,255,0),2)
        cv2.putText(frame,f'{cap_frame}/{number_img_captured}',(20,400),cv2.FONT_HERSHEY_PLAIN,2,(0,255,0),2)

    else:

        if cap_frame>=number_img_captured:
            capture_img=False

        cv2.putText(frame,"Not Capturing",(20,450),cv2.FONT_HERSHEY_PLAIN,2,(0,255,0),2)

    cv2.imshow('frame',frame)

    key=cv2.waitKey(1)&0xFF

    if key==ord('q'):
        break

    if key==ord('c') and not capture_img:
        capture_img=True
        cap_frame=0
        print("Started capturing 20 images")

cap.release()
cv2.destroyAllWindows()
```
