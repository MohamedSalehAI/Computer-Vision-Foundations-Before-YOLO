```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
os.chdir(r"C:\Users\DT1\Desktop\proj-python\Images_Vision")
```

```python
img=np.zeros((512,512),np.uint8)
cv2.rectangle(img,(0,0),(512,256),(255,255,255),-1)
cv2.rectangle(img,(256,256),(512,512),(128,128,128),-1)

plt.hist(img.ravel(),256,[0,256])
plt.show()
cv2.imshow("img",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
bright_img=cv2.imread("bright_image.jpg")
(b,g,r)=cv2.split(bright_img)
plt.hist(b.ravel(),256,[0,256])
plt.hist(g.ravel(),256,[0,256])
plt.hist(r.ravel(),256,[0,256])
plt.xlim([0,256])
plt.ylim([0,10000])
# plt.hist(bright_img.ravel(),256,[0,256])
plt.show()

cv2.imshow("bright_img",bright_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dark_img = cv2.imread("dark_image.jpg")
(b,g,r)=cv2.split(dark_img)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.hist(b.ravel(),256,[0,256],'b')
plt.hist(g.ravel(),256,[0,256],'g')
plt.hist(r.ravel(),256,[0,256],'r')
plt.title("Histogram")
plt.legend(["Blue", "Green", "Red"], loc="upper right")

# plt.hist(dark_img.ravel(),256,[0,256])
for i,color in enumerate(['b','g','r']):
    plt.subplot(1,2,2)
    plt.plot(cv2.calcHist([dark_img],[i],None,[256],[0,256]),color)
plt.title("Histogram")
plt.legend(["Blue", "Green", "Red"], loc="upper right")

plt.xlim([0,256])
plt.ylim([0,10000])
plt.show()
cv2.imshow("Dark Image", dark_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
