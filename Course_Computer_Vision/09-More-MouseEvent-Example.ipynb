# Detect all basic mouse events

```python
import cv2
path=r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_5.jpg"
img = cv2.imread(path,1)

def mouse_event(event, x, y, flags, param):

    if event == cv2.EVENT_LBUTTONDOWN:
        print("Left button DOWN:", x, y)

    elif event == cv2.EVENT_LBUTTONUP:
        print("Left button UP:", x, y)

    elif event == cv2.EVENT_RBUTTONDOWN:
        print("Right button DOWN:", x, y)

    elif event == cv2.EVENT_RBUTTONUP:
        print("Right button UP:", x, y)

    elif event == cv2.EVENT_MBUTTONDOWN:
        print("Middle button DOWN:", x, y)

    elif event == cv2.EVENT_MBUTTONUP:
        print("Middle button UP:", x, y)

    elif event == cv2.EVENT_MOUSEMOVE:
        print("Mouse position:", x, y)


cv2.imshow("Image", img)
cv2.setMouseCallback("Image", mouse_event)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Draw with the mouse 🖱️

```python
import cv2
import numpy as np

canvas = np.zeros((600, 800, 3), dtype=np.uint8)

drawing = False

def draw(event, x, y, flags, param):

    global drawing

    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True

    elif event == cv2.EVENT_MOUSEMOVE:

        if drawing:
            cv2.circle(canvas, (x, y), 5, (255, 255, 255), -1)

    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False


cv2.namedWindow("Drawing")
cv2.setMouseCallback("Drawing", draw)

while True:

    cv2.imshow("Drawing", canvas)

    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break

    elif key == ord("c"):
        canvas[:] = 0

cv2.destroyAllWindows()
```

# Get RGB/BGR pixel value by clicking

```python
import cv2
path=r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_6.jpg"
img = cv2.imread(path)

def get_pixel(event, x, y, flags, param):

    if event == cv2.EVENT_LBUTTONDOWN:

        pixel = img[y, x]

        print("Position:", x, y)
        print("BGR:", pixel)

        b, g, r = pixel

        print("Blue :", b)
        print("Green:", g)
        print("Red  :", r)


cv2.imshow("Image", img)
cv2.setMouseCallback("Image", get_pixel)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Draw a rectangle by dragging

```python
import cv2
path=r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_7.jpg"
img = cv2.imread(path)

start_point = None
drawing = False

def rectangle(event, x, y, flags, param):

    global start_point, drawing, img

    if event == cv2.EVENT_LBUTTONDOWN:

        start_point = (x, y)
        drawing = True

    elif event == cv2.EVENT_MOUSEMOVE:

        if drawing:

            temp = img.copy()

            cv2.rectangle(
                temp,
                start_point,
                (x, y),
                (0, 255, 0),
                2
            )

            cv2.imshow("Image", temp)

    elif event == cv2.EVENT_LBUTTONUP:

        drawing = False

        cv2.rectangle(
            img,
            start_point,
            (x, y),
            (0, 255, 0),
            2
        )


cv2.namedWindow("Image")
cv2.setMouseCallback("Image", rectangle)

while True:

    cv2.imshow("Image", img)

    key = cv2.waitKey(100) & 0xFF

    if key == ord("q"):
        break

cv2.destroyAllWindows()
```
