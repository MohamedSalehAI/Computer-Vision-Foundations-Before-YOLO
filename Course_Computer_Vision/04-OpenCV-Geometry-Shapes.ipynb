# importing libraries

```python
import numpy as np
import cv2
```

# *black image background*

```python
img=np.zeros([840,840,3],dtype=np.uint8)
cv2.imshow("image black",img)
key=cv2.waitKey(0)
if key==ord('q'):
    cv2.destroyAllWindows()
```

# *drawing in blak background*

1. line drawing

```python
mod_img=cv2.line(img,(0,0),(512,512),(0,255,0),10)
cv2.imshow("line image",mod_img)
key=cv2.waitKey(0)
if key==ord('q'):
    cv2.destroyAllWindows()
```

2. rectangle drawing

```python
mod_img=cv2.rectangle(img,(0,0),(512,512),(255,255,0),10)
cv2.imshow("line image",mod_img)
key=cv2.waitKey(0)
if key==ord('q'):
    cv2.destroyAllWindows()
```

3. circle drawing

```python
mod_img=cv2.circle(img,(512,512),50,(0,0,255),10)
cv2.imshow("line image",mod_img)
key=cv2.waitKey(0)
if key==ord('q'):
    cv2.destroyAllWindows()
```

4. ellipse drawing

```python
mod_img=cv2.ellipse(img,(512,512),(100,50),0,0,180,255,-1)
cv2.imshow("line image",mod_img)
key=cv2.waitKey(0)
if key==ord('q'):
    cv2.destroyAllWindows()
```

4. poly drawing

```python
pts = np.array([
    [10, 12],
    [25, 40],
    [40, 45],
    [50, 20]
], np.int32)

mod_img = cv2.polylines(
    img,
    [pts],
    True,
    (255, 255, 50),
    thickness=2
)

cv2.imshow("Line Image", mod_img)

key = cv2.waitKey(0)

if key == ord('q'):
    cv2.destroyAllWindows()
```

5. put text

```python
font=cv2.FONT_HERSHEY_COMPLEX
mod_img=cv2.putText(img,"puttext opencv",(50,150),font,8,(48,152,58),5)
cv2.imshow("Line Image", mod_img)

key = cv2.waitKey(0)

if key == ord('q'):
    cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
path=r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_1.jpg"
# Load image
img = cv2.imread(path)

# Check image
if img is None:
    print("Image not found")
    exit()

# 1. Draw a line
cv2.line(img, (50, 50), (250, 50), (255, 0, 0), 3)

# 2. Draw a rectangle
cv2.rectangle(img, (50, 100), (250, 220), (0, 255, 0), 3)

# 3. Draw a circle
cv2.circle(img, (400, 150), 70, (0, 0, 255), 3)

# 4. Draw a triangle
pts = np.array([
    [300, 300],
    [200, 450],
    [400, 450]
], np.int32)

cv2.polylines(img, [pts], True, (255, 255, 0), 3)

# 5. Draw a filled circle
cv2.circle(img, (550, 150), 50, (255, 0, 255), -1)

# Show image
cv2.imshow("Geometry Shapes", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
