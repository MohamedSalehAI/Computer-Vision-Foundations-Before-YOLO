```python
import numpy as np
import cv2
import os
os.chdir(r"C:\Users\DT1\Desktop\proj-python\Color_Transfer")
```

# directories

```python
input_dir="source_images/"
target_dir="target_images/"
output_dir="result_images/"
```

```python
input_images=os.listdir(input_dir)
```

```python
def get_mean_std(img):
    mean,std=cv2.meanStdDev(img)
    mean=np.hstack(np.around(mean,2))
    std=np.hstack(np.around(std,2))
    return mean,std
```

```python
target_img=cv2.imread(target_dir + "image_2.jpg")
target_img=cv2.cvtColor(target_img, cv2.COLOR_BGR2LAB)
target_mean,target_std=get_mean_std(target_img)

for img in input_images:
    src_img=cv2.imread(input_dir + img)
    src_img=cv2.cvtColor(src_img, cv2.COLOR_BGR2LAB)
    src_mean,src_std=get_mean_std(src_img)
    # print("Source mean:", src_mean)
    # print("Source std:", src_std)
    h,w,c=src_img.shape
    for i in range(h):
        for j in range(w):
            for k in range(c):
                src_img[i,j,k]=(src_img[i,j,k]-src_mean[k])/src_std[k]*target_std[k]+target_mean[k]
    src_img=cv2.cvtColor(src_img, cv2.COLOR_LAB2BGR)
    cv2.imwrite(output_dir + img, src_img)
```

```python
output_images=os.listdir(output_dir)
for img in output_images:
    cv2.imshow(img, cv2.imread(output_dir + img))
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
