```python
import time
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("image_1.jpg")
original_img=img.copy()

width, height = img.size
print("Image size:", width, "x", height)
img_list = img.load()

time_1 = time.time()

for i in range(width):
    for j in range(height):
        img_list[i, j] = (
            img_list[i, j][0] + 150,
            img_list[i, j][1] + 50,
            img_list[i, j][2] + 250
        )

time_2 = time.time()

print("Processing time:", time_2 - time_1, "seconds")
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(original_img)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(img)
plt.title("Processed Image")
plt.axis("off")
plt.tight_layout()
plt.show()
```

#Another way processing more fast by numpy

```python
import time
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = Image.open("image_1.jpg")

# Convert to NumPy array
img_array = np.array(img)

# Keep original image
original = img_array.copy()
print("type of original",type(original))

# Start timing
time_1 = time.perf_counter()

# Increase brightness by 50
processed = np.clip(img_array.astype(np.int16) + 50, 0, 255).astype(np.uint8)

# End timing
time_2 = time.perf_counter()

print("Processing time:", time_2 - time_1, "seconds")

# Show BEFORE and AFTER
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Before")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(processed)
plt.title("After")
plt.axis("off")

plt.show()
```
