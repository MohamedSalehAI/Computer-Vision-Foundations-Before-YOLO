```python
import cv2
import imutils
import datetime
import time
import os
os.chdir(r"D:\proj-python\Videos")
TrDict={'csrt':cv2.TrackerCSRT_create,
        'kcf':cv2.TrackerKCF_create,
        'mil':cv2.TrackerMIL_create,
        # 'boosting':cv2.TrackerBoosting_create,
        # 'medianflow':cv2.TrackerMedianFlow_create,
        # 'mosse':cv2.TrackerMOSSE_create
        }
trackers = cv2.legacy.MultiTracker_create()
```

```python
cap=cv2.VideoCapture(1)
ret,frame=cap.read()
frame=cv2.rotate(frame,cv2.ROTATE_90_CLOCKWISE)
frame=imutils.resize(frame,width=800)
(H,W)=frame.shape[:2]
k=2
for i in range(k):
  cv2.imshow("frame",frame)
  bb=cv2.selectROI("frame",frame,False,False)
  if bb == (0, 0, 0, 0):
        print("ROI not selected")
        continue
  tracker=TrDict['csrt']()
  trackers.add(tracker,frame,bb)
cv2.destroyWindow("frame")
```

```python
fps_start_time = datetime.datetime.now()
fps = 0
total_frames = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    total_frames += 1

    frame = imutils.resize(frame, width=800)
    frame = cv2.rotate(frame, cv2.ROTATE_90_CLOCKWISE)

    success, boxs = trackers.update(frame)

    if not success:
        break

    for box in boxs:
        (x, y, w, h) = [int(v) for v in box]
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0,255,0), 2)

    fps_end_time = datetime.datetime.now()
    time_diff = (fps_end_time - fps_start_time).total_seconds()

    if time_diff >= 0.5:
        fps = total_frames / time_diff
        total_frames = 0
        fps_start_time = fps_end_time

    cv2.putText(frame, "FPS: {:.2f}".format(fps), (75,50), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,0,255), 2)

    cv2.imshow("Frame", frame)

    if cv2.waitKey(5) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
