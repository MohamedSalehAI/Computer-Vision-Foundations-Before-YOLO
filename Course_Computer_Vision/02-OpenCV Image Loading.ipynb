```python
import cv2

# Image path
path = r"C:\Users\DT1\Desktop\proj-python\Images_Vision\image_2.jpg"

# Read image in different modes
img_gray = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
img_bgr = cv2.imread(path, cv2.IMREAD_COLOR)
img_alpha = cv2.imread(path, cv2.IMREAD_UNCHANGED)

images = [img_gray, img_bgr, img_alpha]
titles = ["Gray", "BGR", "Unchanged"]

# Display images
for idx, img in enumerate(images):
    cv2.imshow(titles[idx], img)

# Wait for keyboard input
key = cv2.waitKey(0)

if key == ord("q"):
    cv2.destroyAllWindows()

elif key == ord("s"):
    cv2.imwrite("image_gray.jpg", img_gray)
    cv2.imwrite("image_bgr.jpg", img_bgr)
    cv2.imwrite("image_unchanged.jpg", img_alpha)
    cv2.destroyAllWindows()

elif key == 27:  # ESC
    cv2.destroyAllWindows()
```
