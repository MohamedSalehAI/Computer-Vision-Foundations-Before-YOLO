# Simple Project

```python

cap=cv2.VideoCapture(0)
cap.set(3,1280)
cap.set(4,720)
cap.set
(5,30)

print("Width:",cap.get(3),"Height:",cap.get(4),"FPS:",cap.get(5))

detector=FaceMeshDetector(maxFaces=1)

W=6.3
d=75
f=None

Texts=["FACE DISTANCE SYSTEM","--------------------","Pixel Width:","Focal Length:","Distance:"]

while True:
    ret,frame=cap.read()

    if not ret:
        break

    img,faces=detector.findFaceMesh(frame,draw=True)
    imgtext=np.zeros_like(frame)

    if faces:
        face=faces[0]

        pointLeft=face[145]
        pointRight=face[374]

        cv2.line(img,pointLeft,pointRight,(0,200,0),3)
        cv2.circle(img,pointLeft,5,(255,0,255),cv2.FILLED)
        cv2.circle(img,pointRight,5,(255,0,255),cv2.FILLED)

        w,_=detector.findDistance(pointLeft,pointRight)

        if f is None:
            f=(W*d)/w

        distance=(W*f)/w

         

        cvzone.putTextRect(img,f"Distance: {distance:.1f} cm",(face[10][0]-100,face[10][1]-80),scale=1.5)

        cvzone.putTextRect(img,f"Width: {int(w)} px",(face[10][0]-100,face[10][1]-30),scale=1.2)

        Values=[f"{int(w)} px",f"{f:.2f} px",f"{distance:.1f} cm"]

        for i,txt in enumerate(Texts):
            y=40+i*40
            scale=0.4+(int(w)>50)*0.2
            cv2.putText(imgtext,txt,(20,y),cv2.FONT_HERSHEY_SIMPLEX,scale,(255,255,255),2)

        for i,value in enumerate(Values):
            y=110+i*40
            scale=0.4+(int(w)>50)*0.2
            cv2.putText(imgtext,value,(250,y),cv2.FONT_HERSHEY_SIMPLEX,scale,(255,255,255),2)

    else:
        cv2.putText(imgtext,"NO FACE DETECTED",(20,50),cv2.FONT_HERSHEY_SIMPLEX,0.8,(255,255,255),2)

    cv2.imshow("Face Distance Measurement",img)
    cv2.imshow("System Information",imgtext)

    key=cv2.waitKey(1)&0xFF

    if key==ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python

# Camera
cap=cv2.VideoCapture(1)
cap.set(3,1280)
cap.set(4,720)
cap.set(5,30)

print("Width:",cap.get(3),"Height:",cap.get(4),"FPS:",cap.get(5))

# Face Mesh
detector=FaceMeshDetector(maxFaces=1)

# Calibration
W=6.3
d=75
f=None

# Smoothing
alpha=0.25
smooth_w=None
smooth_d=None
smooth_x=None
smooth_y=None

# Text List
Texts=["FACE DISTANCE SYSTEM","--------------------","Pixel Width:","Focal Length:","Distance:","Scale:"]

while True:
    ret,frame=cap.read()
    frame=cv2.flip(frame,0)
    # frame=cv2.rotate(frame,cv2.ROTATE_90_COUNTERCLOCKWISE)
    if not ret:
        break

    img,faces=detector.findFaceMesh(frame,draw=True)
    imgtext=np.zeros_like(frame)

    if faces:
        face=faces[0]

        pointLeft=face[145]
        pointRight=face[374]

        # Pixel Width
        w,_=detector.findDistance(pointLeft,pointRight)

        if smooth_w is None:
            smooth_w=w
        else:
            smooth_w=alpha*w+(1-alpha)*smooth_w

        # Calibration
        if f is None:
            f=(smooth_w*d)/W

        # Distance
        distance=(W*f)/smooth_w

        if smooth_d is None:
            smooth_d=distance
        else:
            smooth_d=alpha*distance+(1-alpha)*smooth_d

        # Face Position
        x,y=face[10][0],face[10][1]

        if smooth_x is None:
            smooth_x=x
            smooth_y=y
        else:
            smooth_x=alpha*x+(1-alpha)*smooth_x
            smooth_y=alpha*y+(1-alpha)*smooth_y

        # Dynamic Scale
        scale=np.clip(smooth_w/50,0.6,2.2)
        thickness=max(1,int(1.5*scale))
        padding=int(8*scale)

        # Drawing
        cv2.line(img,pointLeft,pointRight,(0,230,0),max(1,int(2*scale)))
        cv2.circle(img,pointLeft,max(2,int(4*scale)),(255,0,255),cv2.FILLED)
        cv2.circle(img,pointRight,max(2,int(4*scale)),(255,0,255),cv2.FILLED)

        # Distance Text
        cvzone.putTextRect(img,f"Distance: {smooth_d:.1f} cm",(int(smooth_x-100*scale),int(smooth_y-70*scale)),scale=scale,thickness=thickness,offset=padding)

        cvzone.putTextRect(img,f"Width: {int(smooth_w)} px",(int(smooth_x-100*scale),int(smooth_y-25*scale)),scale=scale*0.85,thickness=thickness,offset=padding)

        # Text List
        Values=[f"{int(smooth_w)} px",f"{f:.2f} px",f"{smooth_d:.1f} cm",f"{scale:.2f}x"]

        for i,txt in enumerate(Texts):
            y_pos=50+i*45
            cv2.putText(imgtext,txt,(20,y_pos),cv2.FONT_HERSHEY_SIMPLEX,0.65,(255,255,255),2,cv2.LINE_AA)

        for i,value in enumerate(Values):
            y_pos=140+i*45
            cv2.putText(imgtext,value,(270,y_pos),cv2.FONT_HERSHEY_SIMPLEX,0.65,(255,255,255),2,cv2.LINE_AA)

    else:
        smooth_w=None
        smooth_d=None
        smooth_x=None
        smooth_y=None

        cv2.putText(imgtext,"NO FACE DETECTED",(20,50),cv2.FONT_HERSHEY_SIMPLEX,0.8,(255,255,255),2,cv2.LINE_AA)

    cv2.imshow("Face Distance Measurement",img)
    cv2.imshow("System Information",imgtext)

    if cv2.waitKey(1)&0xFF==ord("q"):
        break

cap.release()
cv2.destroyAllWindows()

```
