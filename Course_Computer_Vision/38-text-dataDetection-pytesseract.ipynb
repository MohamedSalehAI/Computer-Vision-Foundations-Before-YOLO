# pytesseract img_to_boxes()

```python
model_path=r"C:\Program Files\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd=model_path
img=cv2.imread("text_recognition2.jpg")
h,w=img.shape[:2]
img=cv2.resize(img,(w*2,h*2))
hImg,wImg=img.shape[:2]
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_blur=cv2.GaussianBlur(img_gray,(3,3),0)
boxes=pytesseract.image_to_boxes(img_blur)
print(boxes)
message=[]
for b in boxes.splitlines():
  print("box",b)
  text,x,y,w,h,page=b.split()
  message.append(text)
  x,y,w,h=int(x),int(y),int(w),int(h)
  cv2.rectangle(img,(x,hImg-y),(w,hImg-h),(0,255,0),3)
  cv2.putText(img,text,(x,hImg-y+25),cv2.FONT_HERSHEY_COMPLEX,1,(50,50,55),2)
print("Full Text is:","".join(message).replace(' ','').strip())
cv2.imshow("img",img)
# cv2.imshow("img_gray",img_gray)
cv2.imshow("img_blur",img_blur)
key = cv2.waitKey(0) & 0xFF
if key == ord('q'):
    cv2.destroyAllWindows()
```

# pytesseract img_to_string

```python
model_path=r"C:\Program Files\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd=model_path
img=cv2.imread("text_recognition2.jpg")
h,w=img.shape[:2]
img=cv2.resize(img,(w*2,h*2))
hImg,wImg=img.shape[:2]
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_blur=cv2.GaussianBlur(img_gray,(3,3),0)
thresh=cv2.adaptiveThreshold(img_blur,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY_INV,11,2)
text=pytesseract.image_to_string(thresh,lang='eng')
 
cv2.imshow("img",img)
# cv2.imshow("img_gray",img_gray)
cv2.imshow("img_blur",img_blur)
cv2.imshow("thresh",thresh)
print(text)
key = cv2.waitKey(0) & 0xFF
if key == ord('q'):
    cv2.destroyAllWindows()
```

# pytesseract img_to_data

```python
model_path=r"C:\Program Files\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd=model_path
img=cv2.imread("text_recognition2.jpg")
h,w=img.shape[:2]
img=cv2.resize(img,(w*2,h*2))
hImg,wImg=img.shape[:2]
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_blur=cv2.GaussianBlur(img_gray,(3,3),0)
thresh=cv2.adaptiveThreshold(img_blur,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY_INV,11,2)
data=pytesseract.image_to_data(thresh,lang='eng',output_type=pytesseract.Output.DICT)
n_boxes=len(data['text'])
full_text=[]
for i in range(n_boxes):
    text=data['text'][i]
    x=data['left'][i]
    y=data['top'][i]
    w=data['width'][i]
    h=data['height'][i]
    confidence=float(data['conf'][i])
    if confidence>50:
        full_text.append(text)
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),2)
        cv2.putText(img,text,(x,y-5),cv2.FONT_HERSHEY_SIMPLEX,1,(100,50,55),2)
        print(f"Text:{text} | X: {x} Y: {y} W: {w} H: {h} Confidence: {confidence:.2f} ")
print(f"Full Text: {' '.join(full_text)}")
cv2.imshow("img",img)
# cv2.imshow("img_gray",img_gray)
# cv2.imshow("img_blur",img_blur)
cv2.imshow("thresh",thresh)
key = cv2.waitKey(0) & 0xFF
if key == ord('q'):
    cv2.destroyAllWindows()
```

# Configurations worth memorizing
| Situation             | Configuration                                                                                       |
| --------------------- | --------------------------------------------------------------------------------------------------- |
| Normal document       | `--psm 3`                                                                                           |
| One block             | `--psm 6`                                                                                           |
| One line              | `--psm 7`                                                                                           |
| One word              | `--psm 8`                                                                                           |
| One character         | `--psm 10`                                                                                          |
| Scattered text        | `--psm 11`                                                                                          |
| Numbers only          | `--psm 7 -c tessedit_char_whitelist=0123456789`                                                     |
| Numbers + letters     | `--psm 7 -c tessedit_char_whitelist=ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789` |
| Preserve spacing      | `--psm 6 -c preserve_interword_spaces=1`                                                            |
| Modern/default engine | `--oem 3`                                                                                           |
| Modern + one block    | `--oem 3 --psm 6`                                                                                   |
