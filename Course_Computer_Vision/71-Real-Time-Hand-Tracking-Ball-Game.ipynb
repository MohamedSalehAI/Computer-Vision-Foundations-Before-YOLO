# basic CVZone HandDetector

```python

import cv2
import cvzone
from cvzone.HandTrackingModule import HandDetector

cap = cv2.VideoCapture(1)
ball = cv2.resize(cv2.imread("ball.png", cv2.IMREAD_UNCHANGED), (50, 50))
detector = HandDetector(detectionCon=0.8, maxHands=2)

while True:
    ret, img = cap.read()
    if not ret:
        break

    hands, img = detector.findHands(img, flipType=True)

    if hands:
        for hand in hands:
            x, y, w, h = hand["bbox"]
            center = hand["center"]
            fingers = detector.fingersUp(hand)
            lmList = hand["lmList"]
            cvzone.putTextRect(img,str(fingers),(x-50,y-50),scale=2,thickness=2,colorR=(0,255,0))
            cvzone.overlayPNG(img,ball,[center[0]-25,center[1]-25])

            print("BBox:", x, y, w, h)
            print("Center:", center)
            print("Fingers:", fingers)
            print("LMList:", lmList)
            print("Name:",hand["type"])

    cv2.imshow("Hand Detector", img)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python

height, width = 1080, 1920

cap = cv2.VideoCapture(1)
cap.set(3, width)
cap.set(4, height)

detector = HandDetector(detectionCon=0.8)

ball = cv2.resize(cv2.imread("ball.png", cv2.IMREAD_UNCHANGED), (50, 50))
speech = cv2.resize(cv2.imread("speech.png", cv2.IMREAD_UNCHANGED), (150, 50))
gameover = cv2.resize(cv2.imread("gameover.png"), (width, height))

position = [50, 50]
speedx, speedy = 10, 10
score = 0

while True:
    ret, img = cap.read()
    if not ret:
        break

    cv2.rectangle(img, (0, 0), (width, height), (0, 0, 255), 8)
    

    hands, img = detector.findHands(img, flipType=True)

    if hands:
        x, y, w, h = hands[0]["bbox"]

        x1 = np.clip(x - speech.shape[1] // 2, 50,width-100)
        img = cvzone.overlayPNG(img, speech, [x1, 600])
        cv2.rectangle(img, (0,650), (width, 600), (0, 255, 255), 6)

        if position[1] < 50:
            speedy = -speedy
            position[1] += 10

        if x1 - 10 < position[0] < x1 + speech.shape[1] and 700 < position[1] < 800:
            speedy = -speedy
            position[1] -= 100
            score += 1

    if position[1] > 900:
        img = gameover

        cvzone.putTextRect(img, f"Final score {score}", [300, 190], 1.9, 2, colorR=(0, 0, 0))
        cvzone.putTextRect(img, "Press R to restart", [290, 305], 1.82, 2, colorR=(0, 0, 0))
        cvzone.putTextRect(img, "Press Esc to quit", [290, 345], 1.82, 2, colorR=(0, 0, 0))

    else:
        if position[0] <= 20 or position[0] >= width - 50:
            speedx = -speedx

        position[0] += speedx
        position[1] += speedy

    cvzone.putTextRect(img, f"Score {score}", [270, 30], 1.5)
    img = cvzone.overlayPNG(img, ball, position)

    cv2.imshow("frame", img)

    key = cv2.waitKey(1)

    if key in [ord("r"), ord("R")]:
        position = [50, 50]
        speedx, speedy = 10, 10
        score = 0

    elif key == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
