# Hough Line transform

```python
img=cv2.imread("pedestrian.jpg")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
edges=cv2.Canny(gray,75,150)
lines=cv2.HoughLinesP(edges,1,np.pi/180,30,maxLineGap=250)
for line in lines:
    x1,y1,x2,y2=line[0]
    cv2.line(img,(x1,y1),(x2,y2),(0,255,0),4)
cv2.imshow("Edges",edges)
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import os
os.chdir(r"D:\proj-python\Videos")

cap=cv2.VideoCapture("Vehicle Dataset Sample 3.mp4")
 
fps = cap.get(cv2.CAP_PROP_FPS)
delay = int((1000 / fps) * 0.5)

while True:
    ret,frame=cap.read()
    if ret==True:
        frame=cv2.GaussianBlur(frame,(5,5),0)
        hsv=cv2.cvtColor(frame,cv2.COLOR_BGR2HSV)
        low_yellow=np.array([15,70,120])
        up_yellow=np.array([120,255,255])
        mask=cv2.inRange(hsv,low_yellow,up_yellow)
        edges=cv2.Canny(mask,75,150)
        lines=cv2.HoughLinesP(edges,1,np.pi/180,50,maxLineGap=250)
        if lines is not None:
          for line in lines:
           x1,y1,x2,y2=line[0]
           cv2.line(frame,(x1,y1),(x2,y2),(0,255,0),4)
        
        cv2.imshow("Edges",edges)
        cv2.imshow("mask",mask)
        cv2.imshow("Image",frame)
        key=cv2.waitKey(delay)
        if key==ord("q"):
            break
    else:
        break
cap.release()
cv2.destroyAllWindows()
```

# Hough Circle Transform

```python
img=cv2.imread("planets.jpg")
cv2.imshow("detected circles",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img=cv2.imread("planets.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_blur = cv2.medianBlur(gray, 5)

circles = cv2.HoughCircles(
    gray_blur,
    cv2.HOUGH_GRADIENT,
    dp=1,
    minDist=100,
    param1=100,
    param2=20,
    minRadius=25,
    maxRadius=80
)

if circles is not None:
    circles = np.uint16(np.around(circles))
    print(circles.shape)
    print("Number of circles:",len(circles[0]))
    for x, y, r in circles[0]:
        cv2.circle(img, (x, y), r, (0, 255, 0), 2)
        cv2.circle(img, (x, y), 2, (0, 0, 255), 3)

cv2.imshow("detected circles", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
