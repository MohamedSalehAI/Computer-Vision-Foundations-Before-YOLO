```python

import cv2
from pyzbar import pyzbar
import time

cap=cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH,980)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT,640)

if not cap.isOpened():
    raise RuntimeError("Could not open camera.")

last_qr=""
capture_time=0

while True:
    ret,frame=cap.read()

    if not ret:
        break

    frame=cv2.flip(frame,1)
    decoded_qr=pyzbar.decode(frame)

    for obj in decoded_qr:
        qr_data=obj.data.decode("utf-8")
        x,y,w,h=obj.rect

        if qr_data!=last_qr:
            print("Type:",obj.type)
            print("Data:",qr_data)
            last_qr=qr_data
            capture_time=time.time()

        cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),2)
        cv2.putText(frame,qr_data,(x,y-10),cv2.FONT_HERSHEY_SIMPLEX,0.7,(0,255,0),2)

        if time.time()-capture_time<10:
            cv2.putText(frame,"QR Captured - Opening link...",(20,40),cv2.FONT_HERSHEY_SIMPLEX,0.7,(0,255,0),2)
           

    cv2.imshow("QR Code Scanner",frame)

    key=cv2.waitKey(1)&0xFF

    if key==ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
