```python
 #load image
img=cv2.imread("balls.jpg")
cv2.imshow("balls", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Convert To HSV

```python
while True:
    hsv=cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    # Define lower and upper red bounds
    lower_red1 = np.array([0, 100, 100])
    upper_red1 = np.array([10, 255, 255])
    
    lower_red2 = np.array([170, 100, 100])
    upper_red2 = np.array([180, 255, 255])
    
    mask1=cv2.inRange(hsv, lower_red1, upper_red1)
    mask2=cv2.inRange(hsv, lower_red2, upper_red2)
    mask=mask1+mask2
    
    result=cv2.bitwise_and(img,img,mask=mask)
    
    cv2.imshow("Original image", img)
    cv2.imshow("HSV", hsv)
    cv2.imshow("Mask", mask)
    cv2.imshow("Result", result)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
cv2.destroyAllWindows()
```

# By using Trackbar instead of constant values

```python
def nothing(value):
    print(value)
img=cv2.imread('balls.jpg')
frame=cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2.namedWindow('Tracking by HSV')
cv2.createTrackbar('LH','Tracking by HSV',0,255,nothing)
cv2.createTrackbar('LS','Tracking by HSV',0,255,nothing)
cv2.createTrackbar('LV','Tracking by HSV',0,255,nothing)
cv2.createTrackbar('UH','Tracking by HSV',255,255,nothing)
cv2.createTrackbar('US','Tracking by HSV',255,255,nothing)
cv2.createTrackbar('UV','Tracking by HSV',255,255,nothing)

while True:
    lh = cv2.getTrackbarPos('LH', 'Tracking by HSV')
    ls = cv2.getTrackbarPos('LS', 'Tracking by HSV')
    lv = cv2.getTrackbarPos('LV', 'Tracking by HSV')
    uh = cv2.getTrackbarPos('UH', 'Tracking by HSV')
    us = cv2.getTrackbarPos('US', 'Tracking by HSV')
    uv = cv2.getTrackbarPos('UV', 'Tracking by HSV')
    
    lower_bound = np.array([lh, ls, lv])
    upper_bound = np.array([uh, us, uv])
    
    mask = cv2.inRange(frame, lower_bound, upper_bound)
    result = cv2.bitwise_and(img, img, mask=mask)
    
    cv2.imshow('Original', img)
    cv2.imshow('HSV', frame)
    cv2.imshow('Mask', mask)
    cv2.imshow('Result', result)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```

# Apply Object Detection on Video

```python
cap=cv2.VideoCapture(0)
h=cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 580)
w=cap.set(cv2.CAP_PROP_FRAME_WIDTH, 720)
fpr=cap.set(cv2.CAP_PROP_FPS, 30)
cv2.namedWindow("Tracking Video")

cv2.createTrackbar("LH", "Tracking Video", 0, 255, nothing)
cv2.createTrackbar("LS", "Tracking Video", 0, 255, nothing)
cv2.createTrackbar("LV", "Tracking Video", 0, 255, nothing)
cv2.createTrackbar("UH", "Tracking Video", 255, 255, nothing)
cv2.createTrackbar("US", "Tracking Video", 255, 255, nothing)
cv2.createTrackbar("UV", "Tracking Video", 255, 255, nothing)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    font=cv2.FONT_HERSHEY_SIMPLEX
    # Colors in BGR
    WHITE  = (255, 255, 255)
    YELLOW = (0, 255, 255)
    GREEN  = (0, 255, 0)
    RED    = (0, 0, 255)

   
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    
    lh = cv2.getTrackbarPos("LH", "Tracking Video")
    ls = cv2.getTrackbarPos("LS", "Tracking Video")
    lv = cv2.getTrackbarPos("LV", "Tracking Video")
    uh = cv2.getTrackbarPos("UH", "Tracking Video")
    us = cv2.getTrackbarPos("US", "Tracking Video")
    uv = cv2.getTrackbarPos("UV", "Tracking Video")
    
    lower_bound = np.array([lh, ls, lv])
    upper_bound = np.array([uh, us, uv])
    
    mask = cv2.inRange(hsv, lower_bound, upper_bound)
    result = cv2.bitwise_and(frame, frame, mask=mask)
    
    # cv2.imshow("Original", frame)
    # cv2.putText(frame, "Press 'q' to quit", (10, 30), font, 1, (255, 255, 255), 2)
    # cv2.imshow("HSV", hsv)
    # cv2.putText(hsv, "Press 'q' to quit", (10, 30), font, 1, (255, 255, 255), 2)
    
    # Mask information
    cv2.putText(mask, "Press 'q' to quit",
                (10, 30), font, 1, GREEN, 2)
    
    cv2.putText(mask, f"Height: {h}",
                (10, 60), font, 1, YELLOW, 2)
    
    cv2.putText(mask, f"Width: {w}",
                (10, 90), font, 1, YELLOW, 2)
    
    cv2.putText(mask, f"FPS: {fpr:.1f}",
                (10, 120), font, 1, RED, 2)
    
    
    # Result information
    cv2.putText(result, "Press 'q' to quit",
                (10, 30), font, 1, GREEN, 2)
    
    cv2.imshow("Mask", mask)
    cv2.imshow("Result", result)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

# Color Recognition

```python
import cv2

cap = cv2.VideoCapture(0)

# Camera settings
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 840)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 680)
cap.set(cv2.CAP_PROP_FPS, 30)

font = cv2.FONT_HERSHEY_SIMPLEX

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Frame dimensions
    h, w = frame.shape[:2]
    cx, cy = w // 2, h // 2

    # Convert to HSV
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Get center pixel
    hue = hsv[cy, cx, 0]
    b, g, r = frame[cy, cx]

  # Detect color using Hue only

    if hue < 5:
        color = "Red"
    
    elif hue < 10:
        color = "Scarlet"
    
    elif hue < 15:
        color = "Vermilion"
    
    elif hue < 20:
        color = "Orange"
    
    elif hue < 25:
        color = "Amber"
    
    elif hue < 30:
        color = "Gold"
    
    elif hue < 35:
        color = "Yellow"
    
    elif hue < 40:
        color = "Lime Yellow"
    
    elif hue < 45:
        color = "Yellow Green"
    
    elif hue < 55:
        color = "Lime"
    
    elif hue < 65:
        color = "Green"
    
    elif hue < 75:
        color = "Emerald"
    
    elif hue < 85:
        color = "Turquoise"
    
    elif hue < 95:
        color = "Cyan"
    
    elif hue < 105:
        color = "Sky Blue"
    
    elif hue < 115:
        color = "Azure"
    
    elif hue < 125:
        color = "Blue"
    
    elif hue < 135:
        color = "Deep Blue"
    
    elif hue < 145:
        color = "Violet"
    
    elif hue < 155:
        color = "Purple"
    
    elif hue < 165:
        color = "Magenta"
    
    elif hue < 170:
        color = "Pink"
    
    else:
        color = "Red"
    
        # Center marker
    cv2.circle(frame, (cx, cy), 20, (255, 255, 255), 2)

    # Color label
    cv2.rectangle(
        frame,
        (cx - 60, cy - 80),
        (cx + 60, cy - 35),
        (30, 30, 30),
        -1
    )

    cv2.putText(
        frame,
        color,
        (cx - 40, cy - 48),
        font,
        0.8,
        (int(b), int(g), int(r)),
        2,
        cv2.LINE_AA
    )

    # Show video
    cv2.imshow("Color Detector", frame)

    # Quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
