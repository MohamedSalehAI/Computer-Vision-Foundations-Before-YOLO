# Scrapping from google

```python
import os
import time
import urllib.request

from selenium import webdriver
from selenium.webdriver.common.by import By


name = "donkey"

try:
    os.mkdir("download_dataset")
except:
    pass

try:
    os.mkdir("download_dataset/" + name)
except:
    pass


driver = webdriver.Chrome()

url = "https://www.bing.com/images/search?q=" + name

driver.get(url)

time.sleep(5)


links = []
x = 1
last_height = 0


def download_image(url, filename):
    try:
        urllib.request.urlretrieve(url, filename)
        return True
    except:
        return False


while x <= 100:

    pictures = driver.find_elements(By.CSS_SELECTOR, "img.mimg")

    print("Images found:", len(pictures))


    for pic in pictures:

        try:

            img_link = pic.get_attribute("src")

            if img_link and img_link.startswith("http") and img_link not in links:

                links.append(img_link)

                file_name = "download_dataset/" + name + "/" + str(x) + ".jpg"

                if download_image(img_link, file_name):

                    print("Downloaded:", x)

                    x += 1

            if x > 100:
                break

        except:
            pass


    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

    time.sleep(3)


    new_height = driver.execute_script("return document.body.scrollHeight")

    print("Height:", new_height)


    if new_height == last_height:
        break

    last_height = new_height


driver.quit()

print("Finished")
print("Images:", x - 1)
```

# Scrapping from instagram

```python

from selenium import webdriver
from selenium.webdriver.common.by import By
import requests
import os
import time

# =========================
# SETTINGS
# =========================

name = input("Enter search name: ")

username_value = "YOUR_USERNAME"
password_value = "YOUR_PASSWORD"

save_dir = f"download_dataset/{name}"
os.makedirs(save_dir, exist_ok=True)

number_of_images = 100

# =========================
# DRIVER
# =========================

driver = webdriver.Chrome()

driver.get("https://www.instagram.com/accounts/login/")
time.sleep(5)

print("Login URL:", driver.current_url)

# =========================
# LOGIN
# =========================

username = driver.find_element(By.NAME, "email")
password = driver.find_element(By.NAME, "pass")

username.send_keys(username_value)
password.send_keys(password_value)

password.submit()

time.sleep(7)

print("After login:", driver.current_url)

# =========================
# MANUAL VERIFICATION
# =========================

input("Complete Instagram verification if needed, then press Enter...")

print("Current URL:", driver.current_url)

# =========================
# CHECK LOGIN
# =========================

if "/accounts/login" in driver.current_url:

    print("Login was not completed.")
    print("Please login manually in the browser.")

    input("After successful login, press Enter...")

print("Current URL:", driver.current_url)

# =========================
# SEARCH
# =========================

driver.get(
    f"https://www.instagram.com/explore/search/keyword/?q={name}"
)

time.sleep(5)

print("Search URL:", driver.current_url)

# =========================
# SCRAPING
# =========================

image_urls = set()
old_count = 0
no_change = 0

while len(image_urls) < number_of_images:

    images = driver.find_elements(By.TAG_NAME, "img")

    for image in images:

        src = image.get_attribute("src")

        if src and src.startswith("http"):
            image_urls.add(src)

        if len(image_urls) >= number_of_images:
            break

    print("Images found:", len(image_urls))

    # Stop if nothing new is being loaded
    if len(image_urls) == old_count:
        no_change += 1
    else:
        no_change = 0

    if no_change >= 3:
        print("No new images are loading.")
        break

    old_count = len(image_urls)

    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight);"
    )

    time.sleep(3)

# =========================
# DOWNLOAD
# =========================

print("Downloading...")

for index, url in enumerate(image_urls):

    try:

        response = requests.get(url, timeout=10)

        if response.status_code == 200:

            file_path = f"{save_dir}/{index}.jpg"

            with open(file_path, "wb") as file:
                file.write(response.content)

            print("Downloaded:", index)

    except Exception as e:

        print("Error:", index, e)

print("Done!")
print("Saved in:", save_dir)

input("Press Enter to close...")

driver.quit()
```
