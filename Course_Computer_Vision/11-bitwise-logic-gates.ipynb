```python
import cv2
import numpy as np


# ============================================================
# Configuration
# ============================================================

SIZE = 512
BACKGROUND = 0
FOREGROUND = 255

FONT = cv2.FONT_HERSHEY_SIMPLEX
FONT_SCALE = 0.8
THICKNESS = 2


# ============================================================
# Helper Functions
# ============================================================

def create_title(image, title):
    """Add a title to an image."""
    
    output = image.copy()

    cv2.putText(
        output,
        title,
        (20, 40),
        FONT,
        FONT_SCALE,
        255,
        THICKNESS,
        cv2.LINE_AA
    )

    return output


def create_binary_images():
    """Create two binary input images."""

    img1 = np.zeros(
        (SIZE, SIZE),
        dtype=np.uint8
    )

    img2 = np.zeros(
        (SIZE, SIZE),
        dtype=np.uint8
    )

    # Input A: Rectangle
    cv2.rectangle(
        img1,
        (100, 100),
        (400, 400),
        FOREGROUND,
        -1
    )

    # Input B: Circle
    cv2.circle(
        img2,
        (256, 256),
        100,
        FOREGROUND,
        -1
    )

    return img1, img2


def create_gate_row(input_a, input_b, result, gate_name):
    """Create one visualization row for a logic gate."""

    input_a = create_title(input_a, "INPUT A")
    input_b = create_title(input_b, "INPUT B")
    result = create_title(result, gate_name)

    return cv2.hconcat([
        input_a,
        input_b,
        result
    ])


# ============================================================
# Create Inputs
# ============================================================

img1, img2 = create_binary_images()


# ============================================================
# Logic Gates
# ============================================================

bit_and = cv2.bitwise_and(img1, img2)

bit_or = cv2.bitwise_or(img1, img2)

bit_xor = cv2.bitwise_xor(img1, img2)

bit_not_a = cv2.bitwise_not(img1)

bit_not_b = cv2.bitwise_not(img2)


# ============================================================
# Difference Image
# ============================================================

difference = cv2.absdiff(img1, img2)


# ============================================================
# Create Visualization Rows
# ============================================================

row_and = create_gate_row(
    img1,
    img2,
    bit_and,
    "AND"
)

row_or = create_gate_row(
    img1,
    img2,
    bit_or,
    "OR"
)

row_xor = create_gate_row(
    img1,
    img2,
    bit_xor,
    "XOR / DIFFERENCE"
)


# NOT gate only needs one input

row_not_a = cv2.hconcat([
    create_title(img1, "INPUT A"),
    create_title(bit_not_a, "NOT A")
])


row_not_b = cv2.hconcat([
    create_title(img2, "INPUT B"),
    create_title(bit_not_b, "NOT B")
])


# ============================================================
# Create Logic Gate Dashboard
# ============================================================

top_section = cv2.vconcat([
    row_and,
    row_or,
    row_xor
])

bottom_section = cv2.vconcat([
    row_not_a,
    row_not_b
])


# Resize sections so the final window is manageable

top_section = cv2.resize(
    top_section,
    (900, 900)
)

bottom_section = cv2.resize(
    bottom_section,
    (900, 600)
)


# ============================================================
# Display Difference Separately
# ============================================================

difference_display = create_title(
    difference,
    "IMAGE DIFFERENCE"
)

difference_display = cv2.resize(
    difference_display,
    (512, 512)
)


# ============================================================
# Display
# ============================================================

cv2.imshow(
    "Logic Gates Dashboard",
    top_section
)

cv2.imshow(
    "NOT Gates",
    bottom_section
)

cv2.imshow(
    "Difference Between Images",
    difference_display
)


# ============================================================
# Exit
# ============================================================

cv2.waitKey(0)
cv2.destroyAllWindows()
```
